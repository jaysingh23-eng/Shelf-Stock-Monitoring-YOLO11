```python
import os
import pandas as pd
from ultralytics import YOLO

model_path = r"C:\Users\singh\Shelf-Stock-Monitoring\runs\detect\shelf_void_detection\weights\best.pt"
test_image_dir = r"C:\Users\singh\Shelf-Stock-Monitoring\dataset\test\images"

model = YOLO(model_path)

images = [
    os.path.join(test_image_dir, f)
    for f in os.listdir(test_image_dir)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

results = model.predict(
    source=images,
    conf=0.25,
    verbose=False
)

data = []

for image_path, result in zip(images, results):

    image_height, image_width = result.orig_shape

    total_void_area = 0

    for box in result.boxes.xyxy:
        x1, y1, x2, y2 = box.tolist()

        width = x2 - x1
        height = y2 - y1

        total_void_area += width * height

    image_area = image_width * image_height

    empty_percentage = (total_void_area / image_area) * 100

    void_count = len(result.boxes)

    if empty_percentage < 10:
        status = "Well Stocked"
    elif empty_percentage < 25:
        status = "Partially Stocked"
    else:
        status = "Needs Restocking"

    data.append({
        "Image": os.path.basename(image_path),
        "Void Count": void_count,
        "Empty Area (%)": round(empty_percentage, 2),
        "Status": status
    })

df = pd.DataFrame(data)

print(df)
```

```python
output_path = r"C:\Users\singh\Shelf-Stock-Monitoring\shelf_monitoring_results.csv"

df.to_csv(output_path, index=False)

print("Results saved to:")
print(output_path)
```

```python
print("Total shelves/images:", len(df))
print("Average empty area:", round(df["Empty Area (%)"].mean(), 2), "%")
print("Total detected voids:", df["Void Count"].sum())

print("\nStock Status:")
print(df["Status"].value_counts())
```
